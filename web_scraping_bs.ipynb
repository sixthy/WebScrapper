```python
from bs4 import BeautifulSoup
import requests
import pandas as pd
import csv 
import matplotlib.pyplot as plt
```

```python
# Scraping para PÁGINA PRINCIPAL
url_main = 'https://books.toscrape.com/index.html'

headers = {'User-Agent' : 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/143.0.0.0 Safari/537.36 OPR/127.0.0.0'}

page_main = requests.get(url_main, headers)
```

```python
soup_main = BeautifulSoup(page_main.text, 'html.parser')
```

```python
titleSoup_main = soup_main.find_all('h3')

titleList_main = [h3.a['title'] for h3 in titleSoup_main]

priceSoup_main = soup_main.find_all('p', class_='price_color')

priceList_main = [float(price.string.replace('Â£', '')) for price in priceSoup_main]

ratingBook = soup_main.find_all('p', class_='star-rating')

ratings = {
    "One" : 1,
    "Two" : 2,
    "Three" : 3,
    "Four" : 4,
    "Five" : 5
}

ratingNumber = [ratings[book['class'][1]] for book in ratingBook]

categoryList_main = ['Main' for _ in titleList_main]
```

```python
bookDict = {'Title': titleList_main,
            'Price': priceList_main,
            'Ratings': ratingNumber,
            'Category': categoryList_main}

bookDF = pd.DataFrame(bookDict)
```

```python
# Scraping para Music
url_Music = 'https://books.toscrape.com/catalogue/category/books/music_14/index.html'

headers = {'User-Agent' : 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/143.0.0.0 Safari/537.36 OPR/127.0.0.0'}

page_Music = requests.get(url_Music, headers)

type(page_Music.text)

soup_Music = BeautifulSoup(page_Music.text, 'html.parser')
```

```python
priceSoup_Music = soup_Music.find_all('p', class_ = 'price_color')

priceList_Music = [float(price.string.replace('Â£', '')) for price in priceSoup_Music]

ratingMusic = soup_Music.find_all('p', class_='star-rating')

ratings = {
    "One" : 1,
    "Two" : 2,
    "Three" : 3,
    "Four" : 4,
    "Five" : 5
}

ratingNumberMusic = [ratings[book['class'][1]] for book in ratingMusic]

categoryList_Music = ['Music' for _ in titleList_Music]

MusicDict = {
    'Title' : titleList_Music,
    'Price' : priceList_Music,
    'Ratings' : ratingNumberMusic,
    'Category' : categoryList_Music
}

MusicDF = pd.DataFrame(MusicDict)
```

```python
all_books  = pd.concat([
    bookDF,
    artDF,
    gamesDF,
    HumorDF,
    MusicDF
]).reset_index(drop=True)

all_books['Price'] = pd.to_numeric(all_books['Price'])

all_books.to_csv('books.csv', index=False)
```

```python
all_books.groupby("Category")["Price"].mean().plot(kind="bar", color="skyblue", edgecolor="black")

plt.title("Average Price by Category", fontsize = 15)
plt.xlabel("Category", fontsize = 12)
plt.ylabel("Price", fontsize = 12)

plt.show()
```
